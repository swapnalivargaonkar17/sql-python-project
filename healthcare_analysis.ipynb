```python
import pandas as pd

patients = pd.read_csv("patients.csv")
appointments = pd.read_csv("appointments.csv")
billing = pd.read_csv("billing.csv")
diagnoses = pd.read_csv("diagnoses.csv")
doctors = pd.read_csv("doctors.csv")

print("All datasets loaded successfully!")
```

```python
print("PATIENTS")
print("Rows and Columns:", patients.shape)
print(patients.columns.tolist())

print("\nAPPOINTMENTS")
print("Rows and Columns:", appointments.shape)
print(appointments.columns.tolist())

print("\nBILLING")
print("Rows and Columns:", billing.shape)
print(billing.columns.tolist())

print("\nDIAGNOSES")
print("Rows and Columns:", diagnoses.shape)
print(diagnoses.columns.tolist())

print("\nDOCTORS")
print("Rows and Columns:", doctors.shape)
print(doctors.columns.tolist())
```

```python
datasets = {
    "Patients": patients,
    "Appointments": appointments,
    "Billing": billing,
    "Diagnoses": diagnoses,
    "Doctors": doctors
}

for name, df in datasets.items():
    print("\n" + "=" * 50)
    print(name)
    print("Duplicate rows:", df.duplicated().sum())
    print("Missing values:")
    print(df.isnull().sum())
```

```python
for name, df in datasets.items():
    print("\n" + "=" * 50)
    print(name)
    print(df.dtypes)
```

```python
patients["registration_date"] = pd.to_datetime(
    patients["registration_date"]
)

appointments["appointment_date"] = pd.to_datetime(
    appointments["appointment_date"]
)

print("Date columns converted successfully!")

print("\nPatients registration date:")
print(patients["registration_date"].dtype)

print("\nAppointments date:")
print(appointments["appointment_date"].dtype)
```

```python
print("===== OVERALL HEALTHCARE KPIs =====")

print("Total Patients:", patients["patient_id"].nunique())
print("Total Doctors:", doctors["doctor_id"].nunique())
print("Total Appointments:", appointments["appointment_id"].nunique())
print("Total Bills:", billing["bill_id"].nunique())

print("Total Billing Amount:", billing["bill_amount"].sum())
print("Total Discount:", billing["discount"].sum())
print("Total Net Revenue:", billing["net_amount"].sum())

print("Average Bill Amount:", round(billing["bill_amount"].mean(), 2))
print("Average Net Amount:", round(billing["net_amount"].mean(), 2))
```

```python
status_analysis = appointments["appointment_status"].value_counts()

print("===== APPOINTMENT STATUS ANALYSIS =====")
print(status_analysis)

print("\nAppointment Status Percentage:")
print(
    round(
        appointments["appointment_status"].value_counts(normalize=True) * 100,
        2
    )
)
```

```python
treatment_analysis = billing.groupby("treatment_type").agg(
    total_bills=("bill_id", "count"),
    total_revenue=("net_amount", "sum"),
    average_bill=("net_amount", "mean")
).sort_values("total_revenue", ascending=False)

print(treatment_analysis)
```

```python
department_analysis = doctors.groupby("department").agg(
    total_doctors=("doctor_id", "count"),
    average_experience=("experience_years", "mean"),
    average_consultation_fee=("consultation_fee", "mean")
).sort_values("total_doctors", ascending=False)

print(department_analysis.round(2))
```

```python
diagnosis_analysis = diagnoses.groupby("diagnosis").agg(
    total_cases=("diagnosis_id", "count")
).sort_values("total_cases", ascending=False)

print("===== TOP DIAGNOSES =====")
print(diagnosis_analysis)
```

```python
severity_analysis = diagnoses["severity"].value_counts()

print("===== DIAGNOSIS SEVERITY =====")
print(severity_analysis)

print("\nSeverity Percentage:")
print(
    round(
        diagnoses["severity"].value_counts(normalize=True) * 100,
        2
    )
)
```

```python
patient_spending = billing.groupby("patient_id").agg(
    total_bills=("bill_id", "count"),
    total_spending=("net_amount", "sum")
).sort_values("total_spending", ascending=False)

print("===== TOP 10 PATIENTS BY SPENDING =====")
print(patient_spending.head(10))
```

```python
doctor_appointments = appointments.groupby("doctor_id").agg(
    total_appointments=("appointment_id", "count")
).sort_values("total_appointments", ascending=False)

doctor_appointments = doctor_appointments.merge(
    doctors[["doctor_id", "doctor_name", "department"]],
    left_index=True,
    right_on="doctor_id"
).set_index("doctor_id")

print("===== TOP 10 DOCTORS BY APPOINTMENTS =====")
print(doctor_appointments.head(10))
```

```python
status_revenue = appointments.merge(
    billing[["appointment_id", "net_amount"]],
    on="appointment_id",
    how="left"
)

status_revenue_analysis = status_revenue.groupby("appointment_status").agg(
    total_appointments=("appointment_id", "count"),
    total_revenue=("net_amount", "sum"),
    average_revenue=("net_amount", "mean")
).sort_values("total_revenue", ascending=False)

print("===== APPOINTMENT STATUS vs REVENUE =====")
print(status_revenue_analysis.round(2))
```
